```python
import numpy as np
import statistics

# -----------------------------
# Sample Data
# -----------------------------
data = [10, 12, 14, 16, 1000]   # 1000 is an outlier

print("Data:", data)
```

```python
# -----------------------------
# 1) Mean
# -----------------------------
mean_value = np.mean(data)
print("\n1) Mean:", mean_value)

# -----------------------------
# 2) Weighted Mean
# -----------------------------
values = np.array([80, 70, 90])
weights = np.array([3, 2, 5])

weighted_mean = np.sum(values * weights) / np.sum(weights)
print("\n2) Weighted Mean:", weighted_mean)
```

```python
# -----------------------------
# 3) Median
# -----------------------------
median_value = np.median(data)
print("\n3) Median:", median_value)

# -----------------------------
# 4) Percentile
# -----------------------------
p50 = np.percentile(data, 50)  # same as median
p80 = np.percentile(data, 80)

print("\n4) Percentile:")
print("   50th percentile (Median):", p50)
print("   80th percentile:", p80)
```

```python
# -----------------------------
# 5) Weighted Median (custom function)
# -----------------------------
def weighted_median(values, weights):
    values = np.array(values)
    weights = np.array(weights)

    # Sort values and weights according to values
    sorted_indices = np.argsort(values)
    values_sorted = values[sorted_indices]
    weights_sorted = weights[sorted_indices]

    # Cumulative sum of weights
    cum_weights = np.cumsum(weights_sorted)
    half_weight = np.sum(weights_sorted) / 2

    # Find where cumulative weight crosses half
    for i in range(len(values_sorted)):
        if cum_weights[i] >= half_weight:
            return values_sorted[i]

wm_values = [10, 20, 30]
wm_weights = [1, 2, 7]

print("\n5) Weighted Median:", weighted_median(wm_values, wm_weights))
```

```python
# -----------------------------
# 6) Trimmed Mean
# -----------------------------
def trimmed_mean(data, trim_count=1):
    sorted_data = sorted(data)
    trimmed_data = sorted_data[trim_count: len(sorted_data) - trim_count]
    return np.mean(trimmed_data)

trim_mean = trimmed_mean(data, trim_count=1)
print("\n6) Trimmed Mean (after removing 1 smallest & 1 largest):", trim_mean)
```

```python
# -----------------------------
# 7) Robust (Median is robust)
# -----------------------------
print("\n7) Robust Measure Example:")
print("   Mean (not robust):", mean_value)
print("   Median (robust):", median_value)
```

```python
# 8) Outlier Detection (Simple method)
# Using mean and standard deviation
# -----------------------------
mean_val = np.mean(data)
std_val = np.std(data)
print("Mean Value", mean_val)
print("Standard Devation", std_val)
outliers = []
for x in data:
    if abs(x - mean_val) > 2 * std_val:
        outliers.append(x)

print("\n8) Outlier(s) detected (2*std rule):", outliers)
```

```python
import numpy as np

data = [10, 12, 14, 16, 1000]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
print("Outliers =", outliers)
```
