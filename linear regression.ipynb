```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# linear Regression
X= np.random.rand(100,1)*10
y= 3*X.squeeze()+5+ np.random.randn(100) 
```

```python
model_lr = LinearRegression()
```

```python
model_lr.fit(X,y)
```

```python
y_pred = model_lr.predict(X)
```

```python
y_pred
```

```python
print('linear Regression')
print('MSE:', mean_squared_error(y,y_pred))
print("R2 Scrore:", r2_score(y,y_pred))
```

```python
# multiple Linear Regression

X_multi = np.random.rand(100,2)*10
Y_muulti = 2*X_multi[:,0]+4*X_multi[:,1]+3+np.random.randn(100)
```

```python
model_m = LinearRegression()
```

```python
model_m.fit(X_multi,Y_muulti)
```

```python
Y_pred1 = model_m.predict(X_multi)
```

```python
print(' multiple linear Regression')
print('MSE:', mean_squared_error(Y_muulti,Y_pred1))
print("R2 Scrore:", r2_score(Y_muulti,Y_pred1))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = pd.DataFrame({
    'Model': ['Linear Regression', 'Multiple Linear Regression'],
    'MSE': [1.2857, 0.7549],
    'R2 Score': [0.9841, 0.9956]
})
```

```python
sns.set(style="whitegrid")

plt.figure(figsize=(10,5))

# 🔹 MSE Graph
plt.subplot(1,2,1)
sns.barplot(x='Model', y='MSE', data=data)
plt.title('MSE Comparison')

# 🔹 R2 Graph
plt.subplot(1,2,2)
sns.barplot(x='Model', y='R2 Score', data=data)
plt.title('R2 Score Comparison')

plt.tight_layout()
plt.show()
```
