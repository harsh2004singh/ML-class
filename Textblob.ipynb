```python
from textblob import TextBlob
```

```python
reviews = [
    "The hotel room was spacious and very clean.",
    "Excellent service and friendly staff.",
    "The food was terrible and the room was dirty.",
    "Average stay, nothing special.",
    "I loved the ambiance and the hospitality.",
    "The washroom was not clean and the bed was uncomfortable.",
    "Fantastic location and amazing breakfast.",
    "Very noisy at night and poor room service.",
    "The experience was okay, not too bad.",
    "One of the best hotels I have ever visited.",
    "The staff made us feel welcome and took care of us",
    " Had a great time and would surely return for amazing hospitality.",
    "NOT so Good Staff and not clean Room"
]

df = pd.DataFrame({"Review": reviews})
```

```python
def get_textblob_sentiment(review):
    polarity = TextBlob(review).sentiment.polarity
    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"


df["Sentiment"] = df["Review"].apply(get_textblob_sentiment)
```

```python
print(df)
print("\nSentiment Counts:")
print(df["Sentiment"].value_counts())
```

```python
df = data.drop(columns=['id', 'dateAdded', 'dateUpdated', 'address', 'categories',
       'primaryCategories', 'city', 'country', 'keys', 'latitude', 'longitude',
       'name', 'postalCode', 'province', 'reviews.date', 'reviews.dateSeen',
       'reviews.sourceURLs', 'reviews.rating', 'reviews.title',
       'reviews.userCity', 'reviews.userProvince', 'reviews.username',
       'sourceURLs', 'websites'])
```

```python
df['reviews.text'] = df['reviews.text'].fillna('')
df['reviews.text'] = df['reviews.text'].astype(str)
```

```python
def get_textblob_sentiment(review):
    polarity = TextBlob(review).sentiment.polarity
    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["reviews.text"].apply(get_textblob_sentiment)
```

```python
print(df)
print("\nSentiment Counts:")
print(df["Sentiment"].value_counts())
```
